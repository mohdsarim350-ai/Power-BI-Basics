## Question 1: What is Power BI and why is it used in business?

### Answer

**Microsoft Power BI** is a business intelligence and data visualization platform used to connect to data, transform and model it, create interactive reports and dashboards, and share insights.

Businesses use Power BI to:

- Combine data from different sources.
- Clean and transform data using Power Query.
- Create relationships and data models.
- Build interactive charts, tables, KPIs, and dashboards.
- Track business performance and trends.
- Support data-driven decision-making.
- Share reports with teams and stakeholders.

For example, a real-estate company can use Power BI to analyse property prices, locations, property types, sales trends, and owner information in an interactive report.

## Question 2: Name and explain the three main components of Power BI.

### Answer

The three main components commonly used in the Power BI workflow are:

1. **Power Query** — Used for connecting to data sources and performing data extraction, cleaning, and transformation. Examples include removing duplicates, changing data types, filtering rows, and splitting columns.

2. **Power BI Desktop** — The main development application where data is loaded, relationships are created, calculations are written using DAX, and interactive reports are designed.

3. **Power BI Service** — The cloud platform used to publish, share, collaborate on, and manage Power BI reports and dashboards.

Together, these components support the process of preparing data, modelling it, creating visualizations, and sharing business insights.

## Question 3: Explain the Power BI workflow.

### Answer

A typical Power BI workflow is:

**Connect → Transform → Load → Model → Visualize → Publish/Share**

1. **Connect:** Connect Power BI to sources such as Excel, CSV, SQL databases, web services, or cloud applications.
2. **Transform:** Use Power Query to clean and transform the data.
3. **Load:** Load the prepared data into the Power BI data model.
4. **Model:** Create relationships between tables and define calculated columns/measures where required.
5. **Visualize:** Build charts, tables, cards, slicers, and other report visuals.
6. **Publish:** Publish the report to Power BI Service.
7. **Share and monitor:** Share dashboards/reports with authorized users and maintain scheduled refreshes when required.

## Question 4: List any four data cleaning tasks that can be performed in Power Query.

### Answer

Four common Power Query cleaning tasks are:

1. **Remove duplicates** — Removes repeated records.
2. **Handle missing values** — Replace, fill, or filter null/blank values where appropriate.
3. **Change data types** — Convert columns to the correct type, such as Text, Whole Number, Decimal, or Date.
4. **Split or merge columns** — For example, split `Owner Name` into first and last name.

Other Power Query operations include filtering rows, replacing values, trimming/cleaning text, renaming columns, removing unnecessary columns, and grouping data.

## Question 6: Define Data View, Report View, and Model View. Explain the purpose of each.

### Answer

### 1. Data View

**Data View** displays the rows and columns of the loaded data.

**Purpose:** It is useful for inspecting records, checking column values, and reviewing calculated columns.

### 2. Report View

**Report View** is where interactive reports and visualizations are created.

**Purpose:** It is used to add charts, tables, cards, slicers, filters, KPIs, and other visuals to communicate insights.

### 3. Model View

**Model View** displays the tables and relationships in the data model.

**Purpose:** It is used to create and inspect relationships, understand the data model structure, and manage how tables connect to each other.

In short:

**Data View = inspect data**  
**Report View = build visuals**  
**Model View = manage relationships/model**

## Question 8: Split Owner Name to create two new columns as First Name and Last Name.

### Answer — Power Query method

Assume the dataset contains an `Owner Name` column.

1. Open **Transform Data** to enter Power Query.
2. Select the **Owner Name** column.
3. Go to **Transform → Split Column → By Delimiter**.
4. Select **Space** as the delimiter.
5. Choose the option to split at the **left-most delimiter** if names contain more than two parts and the assignment expects the first word as First Name.
6. Rename the two resulting columns:
   - `Owner Name.1` → **First Name**
   - `Owner Name.2` → **Last Name**
7. Check the resulting values.
8. Click **Close & Apply**.

### Example

| Owner Name | First Name | Last Name |
|---|---|---|
| Rahul Sharma | Rahul | Sharma |
| Anjali Verma | Anjali | Verma |
| Amit Kumar | Amit | Kumar |

### Power Query (M) example

If the exact column is named `Owner Name`, the transformation can also be represented using:

```text
Table.SplitColumn(
    Source,
    "Owner Name",
    Splitter.SplitTextByDelimiter(" ", QuoteStyle.Csv),
    {"First Name", "Last Name"}
)
```

The delimiter and split option should be adjusted if the real-estate dataset contains multi-word names.
